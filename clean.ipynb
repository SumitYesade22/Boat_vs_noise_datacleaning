```python
import re
import pandas as pd

# NOTE: assumes your working dataframe is called `merged`.
# If yours has a different name, rename it below or find-and-replace.

def safe(x):
    return "" if pd.isna(x) else str(x)

# Combine all text sources that might mention driver size
driver_text_blob = (
    merged['product_name'].apply(safe) + " " +
    merged['description_raw_specs'].apply(safe) + " " +
    merged['features_raw_specs'].apply(safe) + " " +
    merged['tags_raw_specs'].apply(safe) + " " +
    merged['ocr_text_raw'].apply(safe) + " " +
    merged['description_raw_ocr'].apply(safe) + " " +
    merged['features_raw_ocr'].apply(safe)
)

def extract_driver_strict(text):
    """
    Find all '<number>mm' mentions where the word 'driver' appears
    within ~25 characters of the number - this avoids false positives
    from unrelated mm mentions (cable length, case dimensions, etc).
    Returns a list of all matches found (a product can have dual drivers).
    """
    results = []
    for match in re.finditer(r'(\d+(?:\.\d+)?)\s*mm', text, re.IGNORECASE):
        start, end = match.span()
        context = text[max(0, start - 25):end + 25].lower()
        if 'driver' in context:
            results.append(float(match.group(1)))
    return results

merged['driver_extracted_list'] = driver_text_blob.apply(extract_driver_strict)
merged['driver_extracted_primary'] = merged['driver_extracted_list'].apply(
    lambda x: x[0] if x else None
)

# Keep the old column for comparison, then fill gaps in driver_size_mm
# with the newly extracted value (don't overwrite values that already exist)
merged['driver_size_mm_old'] = merged['driver_size_mm']
merged['driver_size_mm'] = merged['driver_size_mm'].fillna(merged['driver_extracted_primary'])

# ------------------------------------------------------------------
# Report results
# ------------------------------------------------------------------
old_found = merged['driver_size_mm_old'].notna().sum()
new_found = merged['driver_size_mm'].notna().sum()
total = len(merged)

print(f"Original driver_size_mm coverage: {old_found}/{total} = {old_found/total*100:.1f}%")
print(f"Coverage after text extraction:   {new_found}/{total} = {new_found/total*100:.1f}%")
print(f"Newly filled from text: {new_found - old_found} rows")
print()
print("Value counts after fill:")
print(merged['driver_size_mm'].value_counts(dropna=False).sort_index())

# Optional: inspect exactly which rows got newly filled, to spot-check
newly_filled = merged[merged['driver_size_mm_old'].isna() & merged['driver_size_mm'].notna()]
print(f"\n{len(newly_filled)} rows newly filled - spot check a few:")
print(newly_filled[['product_name', 'driver_size_mm']].head(10).to_string())
```

```python
import re

def get_base_model(x):

    x = str(x).lower().strip()

    # Brand already removed
    x = re.sub(r'^(?:boat|noise)\s+', '', x)

    # Remove only clear descriptive suffixes
    x = re.sub(
        r'\s+(?:'
        r'wireless|'
        r'bluetooth|'
        r'truly wireless|'
        r'truly|'
        r'true wireless|'
        r'true|'
        r'bt|'
        r'open ear|'
        r'ows open ear|'
        r'in ear|'
        r'gaming|'
        r'made in india'
        r')\b.*$',
        '',
        x
    )

    # Remove collaboration / special-edition descriptors
    x = re.sub(
        r'\s+(?:'
        r'batman|superman|captain america|black panther|'
        r'wonder woman|iron man|deadpool|thor|'
        r'kkr|rcb|naruto|sunburn|'
        r'dhruv kapoor|blue tokai|'
        r'netflix stream|'
        r'ondc partner|'
        r'forest green'
        r')'
        r'(?:\s+(?:dc|series|edition|exclusive|online|partner))?'
        r'(?:\s+edition)?'
        r'.*$',
        '',
        x
    )

    x = re.sub(r'\s+', ' ', x).strip()

    return x


merged['base_model'] = merged['base_product'].apply(get_base_model)
```

```python
import re

def clean_base_model(x):

    x = str(x).lower().strip()

    # --------------------------------
    # 1. Hyphen -> space
    # --------------------------------
    x = x.replace('-', ' ')

    # --------------------------------
    # 2. Remove brand
    # --------------------------------
    x = re.sub(r'^(?:boat|noise)\s+', '', x)

    # --------------------------------
    # 3. Remove URL garbage
    # --------------------------------
    x = re.sub(
        r'^https?\s+www\s+boat\s+lifestyle\s+com\s+products\s+boat\s+',
        '',
        x
    )

    # --------------------------------
    # 4. Remove special editions /
    #    collaborations
    # --------------------------------
    x = re.sub(
        r'\s+(?:'
        r'batman|superman|captain america|black panther|'
        r'wonder woman|iron man|deadpool|thor|'
        r'kkr|rcb|naruto|sunburn|'
        r'dhruv kapoor|blue tokai|'
        r'netflix stream|telegram|'
        r'forest green'
        r')'
        r'(?:\s+(?:dc|series))?'
        r'(?:\s+edition)?'
        r'(?:\s+exclusive)?'
        r'.*$',
        '',
        x
    )

    # --------------------------------
    # 5. Remove descriptive suffix
    # --------------------------------
    x = re.sub(
        r'\s+(?:'
        r'made in india|'
        r'long playback|'
        r'large playback|'
        r'noise cancelling|'
        r'active noise cancellation|'
        r'with .*|'
        r'bluetooth|'
        r'wireless|'
        r'truly wireless|'
        r'true wireless|'
        r'truly|'
        r'true|'
        r'bt|'
        r'gaming|'
        r'earbuds|'
        r'earphones|'
        r'earphone|'
        r'headphones|'
        r'headphone|'
        r'neckband|'
        r'in ear|'
        r'open ear|'
        r'wired'
        r')\b.*$',
        '',
        x
    )

    # --------------------------------
    # 6. Clean remaining spaces
    # --------------------------------
    x = re.sub(r'\s+', ' ', x).strip()

    return x


merged['base_model'] = merged['base_model'].apply(clean_base_model)
```

```python
import re

def clean_edition(x):
    x = str(x).lower().strip()

    # edition-related suffixes remove
    x = re.sub(
        r'\s+(?:dc|ss|special|limited)\s+edition\b.*$',
        '',
        x
    )

    # standalone edition bhi remove
    x = re.sub(
        r'\s+edition\b.*$',
        '',
        x
    )

    return re.sub(r'\s+', ' ', x).strip()

merged['base_model'] = merged['base_model'].apply(clean_edition)
```
